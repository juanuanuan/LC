## Bin Packing
---

```python

from ortools.linear_solver import pywraplp


def binpack(C, W):
    s = pywraplp.Solver.CreateSolver('SCIP')
    N = len(W)

    A = [[s.BoolVar(f'A[{i}, {j}]') for j in range (N)] for i in range(N)]
    y = [s.BoolVar(f'y[{j}]') for j in range(N)]

    for i  in range(N):
        s.Add(sum(A[i][j] for j in range(N)))
    for j in range(N):
        s.Add(sum(W[i] * A[i][j] for i in range(N)) <= C * y[j])

    s.Minimize(sum(y[j] for j in range(N)))

    status = s.Solve()
    if status == pywraplp.Solver.OPTIMAL or status == pywraplp.Solver.FEASIBLE:
        return #por completar
    else:
        return []
#assert ...
```

## KnapSack

---

```python
def knap(C, I):
    s = pywraplp.Solver.CreateSolver('SCIP')
    N = len(I)
    W = [w for (w,v) in I]
    V = [v for (w,v) in I]

    x = [s.BoolVar(f'x[{i}]') for i in range(N)]

    s.Add(sum(W[i] * x[i] for i in range(N)) <= C)

    s.Maximize(sum(V[i] * x[i] for i in range(N)))

    status = s.Solve()
    if status == pywraplp.Solver.OPTIMAL or status == pywraplp.Solver.FEASIBLE:
        chosen = [i for i in range(N) if int(x[i].solution_value()) == 1]
        value = sum(V[i] for i in chosen)
        weight = sum(W[i] for i in chosen)
        print("Value: ", value)
        print("Weight: ", weight)
        return chosen
    else:
        return []
assert knap(15, [(12,4), (2,2), (1,2), (1,1), (4,10)]) == [1,2,3, 4]
```
